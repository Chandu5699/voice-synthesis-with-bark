## Basics

```python
from bark import SAMPLE_RATE, generate_audio, preload_models
from IPython.display import Audio

preload_models()
```

```python
text_prompt = """
     Hello, this is Balaji from Mitra Robot. And, uh — and I like to build robots. [laughs]
     But I also have other interests such as biking.
"""
audio_array = generate_audio(text_prompt)
Audio(audio_array, rate=SAMPLE_RATE)
```

## Advanced examples

```python
text_prompt = """
    नमस्ते, मैं मित्रा रोबोट से बालाजी हूं। और, उह - और मुझे रोबोट बनाना पसंद है। [हँसते हुए]
     लेकिन मेरी अन्य रुचियाँ भी हैं जैसे बाइक चलाना।
"""
audio_array = generate_audio(text_prompt)
Audio(audio_array, rate=SAMPLE_RATE)
```

```python
text_prompt = """
    WOMAN: I would like an oatmilk latte please.
    MAN: Wow, that's expensive!
"""
audio_array = generate_audio(text_prompt)
Audio(audio_array, rate=SAMPLE_RATE)
```

```python
text_prompt = """
    I have a silky smooth voice, and today I will tell you
    about the exercise regimen of the common sloth.
"""
audio_array = generate_audio(text_prompt, history_prompt="v2/en_speaker_1")
Audio(audio_array, rate=SAMPLE_RATE)
```
